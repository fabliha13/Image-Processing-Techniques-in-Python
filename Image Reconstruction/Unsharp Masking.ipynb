Unsharp masking is an image processing technique used to sharpen an image by enhancing its edges and making details stand out more clearly. Despite the name, "unsharp" masking actually makes images appear sharper. It’s commonly used in photography, printing, and medical imaging to increase the clarity and detail of an image.

How Unsharp Masking Works?

Blur the Image:
A slightly blurred (low-pass filtered) version of the original image is created, often using a Gaussian blur.

Subtract the Blurred Image from the Original:
The blurred image is subtracted from the original image, producing an image that contains only the high-frequency details (edges and fine details). This result is known as the "mask."

Add the Mask Back to the Original Image:
The mask is then added back to the original image with a scaling factor, which controls the sharpness. This enhances the edges and makes the image appear sharper.
